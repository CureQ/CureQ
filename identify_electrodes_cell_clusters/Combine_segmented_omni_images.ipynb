```python
import os
import shutil
import matplotlib.image as mpimg
import matplotlib.pyplot as plt

"""
With this code pictures can be combined
This code block was used to combine Segmentationplots till one plot.
"""
#Give here the path to your map
source = r'Paste your path here'        #The r stands for a read string. This is needed because of the \ in a path
#Give here the amount of elektrodes
electrode=16    #The code works now only with 16 electrodes it is hard encoded

#This function loads the pictures and retracts the names
def load_pictures(source, electrode):
    names=[]
    pics=[]
    #Retracts all the files from the map
    for files in os.listdir(source):
        names.append(files)
    #Creates a matrix with the right amount of columns and rows    
    row= [names[i:i+4] for i in range(0, len(names), 4)]
    """This matrix is sorted in reverse because the electrodes in Axion Biosystems are counted from left to right from bottom to top.
        The files are named after the position in the well."""
    sorting= sorted(row, reverse=True)
    #This for loop finds all the pictures in this map
    for element in sorting:
        for image in element:
            print(image)
            image_path = os.path.join(source, image)
            img = mpimg.imread(image_path)
            pics.append(img)
            electrode-=1
    return pics, sorting

#This is the function that creates the picture
def same_pic(pics, names, outputfile):
    row_amount = 4
    col_amount = 4
    print(f"rows: {row_amount}, cols: {col_amount}")
    # Create picture of all images combined
    #Here the plot is created. With row_amount and col_amount rows and columns.
    #The size of the plot is set at 8 by 8 inch. It is set that there should be no space between the subplots.
    fig, axs = plt.subplots(row_amount ,col_amount, figsize=(8, 8), dpi= 100, gridspec_kw=dict(wspace=0, hspace=0))
    axs = axs.ravel()
    plt.subplots_adjust(left=0, right=1, top=1, bottom=0)
    for i, ax in enumerate(axs):
        img = pics[i]
        well = ax.imshow(img, cmap="gray")
        ax.set_xticks([])
        ax.set_yticks([])
    
    
    fig.patch.set_facecolor('none')
    fig.subplots_adjust(wspace=0, hspace=0)
    fig.savefig(outputfile, dpi=100, bbox_inches='tight', pad_inches=0)
    print("Image saved!")

#calling the functions
pics, names = load_pictures(source, electrode=16)
#here i used the outputpath + how i want this picture to be called
same_pic(pics, names, r"Give here your path\here should come the name of your new picture.png")
```
