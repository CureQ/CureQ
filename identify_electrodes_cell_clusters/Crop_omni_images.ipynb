```python
from PIL import Image
import glob
import os

#Input map
input_folder = r"path to map with input pictures"
#output map
output_folder= r"path to the map where i want the output pictures to be stored in"
#if you want to set the box valuas by hand you should use the code below
# # crop-coördinaten invoeren
# print("Voer de crop-coördinaten in (in pixels):")
# x1 = int(input("x1: "))
# y1 = int(input("y1: "))
# x2 = int(input("x2: "))
# y2 = int(input("y2: "))

#The measurements for the box of the signals is:
# crop_box = (270, 150, 2600, 1000)

#Measurements of the well cut out that were used
#(lowest x-values, lowest y-values, highest x-value, highest y-value)
crop_box= (1450,1350,2650,2600)

#Puts all output files in the same map
pattern = os.path.join(input_folder, "*.jpeg")

for file in glob.glob(pattern):
    img = Image.open(file)
    cropped = img.crop(crop_box)
    basename = os.path.basename(file)
    save_path = os.path.join(output_folder, f"cropped_{basename}")
    print(cropped)
    cropped.save(save_path)
    print(f"Opgeslagen: {save_path}")
```
